##3. Asumsi Jumlah Unit dan Ukuran Rata-rata
Perkiraan/asumsi jumlah file:
1. Teks: 500 Dokumen
2. Gambar: 300 Foto
3. Audio: 10 Rekaman
4. Video: 5 Video
5. Spasial: 50 Data lokasi

```python
data_storage = pd.DataFrame({
    "Jenis Data": [
        "Teks",
        "Gambar",
        "Audio",
        "Video",
        "Spasial"
    ],
    "Jumlah Unit": [
        500,
        300,
        10,
        5,
        50
    ],
    "Ukuran per Unit (MB)": [
        10 / 1024,
        3,
        50,
        1024,
        0.5 / 1024
    ]
})
data_storage["Total (MB)"] = (
    data_storage["Jumlah Unit"] * data_storage["Ukuran per Unit (MB)"]
)

display(data_storage.round(2))

total_mb = data_storage["Total (MB)"].sum()
total_gb = total_mb / 1024

print("Total penyimpanan bulan pertama:")
print(round(total_mb, 2), "MB")
print(round(total_gb, 2), "GB")
```

##4. Estimasi Total Storage + Cadangan 30%

```python
cadangan = 0.30
cadangan_gb = total_gb * cadangan
total_cadangan = total_gb + cadangan_gb

print("Total Kebutuhan:")
print(round(total_gb, 2), "GB")
print("Total Kebutuhan dengan Cadangan 30%:")
print(round(total_cadangan, 2), "GB")
```

##5. Estimasi Storage 1 Tahun dengan Asumsi Pertumbuhan 10% Per Bulan

```python
pertumbuhan = 0.10
total_sementara = total_gb
total_tahunan = 0

for bulan in range(1, 13):
    total_tahunan = total_tahunan + total_sementara
    print("Bulan", bulan, ":", round(total_sementara, 2), "GB")
    total_sementara = total_sementara + (total_sementara * pertumbuhan)

print("Total kebutuhan selama 1 tahun:", round(total_tahunan, 2), "GB")

cadangan_tahunan = total_tahunan + (total_tahunan * cadangan)
print("Total kebutuhan 1 tahun + cadangan 30%:", round(cadangan_tahunan, 2), "GB")
```
