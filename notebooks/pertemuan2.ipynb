**Identifikasi Dataset Publik**

Nama dataset: Jumlah Remaja Berdasarkan Desa/Kelurahan di Jawa Barat

Sumber: Satu Data Indonesia

Link dataset: https://data.go.id/dataset/dataset/jumlah-remaja-berdasarkan-desa-kelurahan-di-jawa-barat

Format: CSV

Jenis data: Structured (data terstruktur)

Saya memilih dataset “Jumlah Remaja Berdasarkan Desa/Kelurahan di Jawa Barat” dari Portal Satu Data Indonesia. Dataset ini berisi informasi mengenai jumlah remaja berdasarkan desa atau kelurahan di Provinsi Jawa Barat. Dataset ini termasuk data terstruktur karena disajikan dalam format tabel yang memiliki baris dan kolom dengan atribut tertentu. Setiap baris mewakili catatan data suatu wilayah, sedangkan kolom berisi informasi yang menjelaskan wilayah dan jumlah remaja. Jenis data ini relatif mudah diolah menggunakan Python, misalnya dengan library pandas melalui fungsi pd.read_csv() untuk berkas CSV. Data ini dapat digunakan untuk mempelajari persebaran remaja di berbagai wilayah serta menjadi bahan analisis kependudukan.

**Penyimpanan JSON Bersarang**

JSON bersarang tidak dapat langsung dimasukkan ke dalam tabel relasional biasa tanpa penyesuaian karena data JSON dapat memiliki objek dan daftar di dalam suatu kolom. JSON itu seperti kotak yang berisi kotak-kotak lain, sedangkan tabel hanya punya baris dan kolom, dan setiap kotaknya hanya boleh berisi satu isi sederhana. Contohnya, pada berkas ulasan.json, sebuah objek ulasan dapat berisi rating dan komentar, sedangkan fasilitas dapat berisi daftar beberapa fasilitas. Struktur tersebut tidak otomatis menjadi kolom-kolom terpisah dalam tabel relasional. Agar lebih mudah diolah menggunakan SQL, data perlu ditransformasikan terlebih dahulu. Misalnya, informasi destinasi disimpan dalam tabel destinasi, ulasan disimpan dalam tabel ulasan, dan daftar fasilitas disimpan dalam tabel fasilitas. Tabel-tabel tersebut kemudian dapat dihubungkan menggunakan kunci yang sesuai.
